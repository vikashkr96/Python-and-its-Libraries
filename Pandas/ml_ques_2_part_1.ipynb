```python
import numpy as np
import pandas as pd

# a) Create a Series using pandas and display


print("creating series from a Numpy N-D Array")
arr = np.array([3,5,6,7,8,0,9,67,45])

series1 = pd.Series(arr)
print(series1)

print("creating series from a python dictionary")
dic = {"a":1, "b":2, "c":3, "d":4, "e":5}
series2 = pd.Series(dic)
print(series2)
```

```python
# b) Access the index and the values of our Series

dic = {"a":1, "b":2, "c":3, "d":4, "e":5}
series2 = pd.Series(dic)

print(series2.index)
print("Values : ",series2.values)
```

```python
# c) Compare an array using Numpy with a series using pandas

# A NumPy array is a homogeneous n-dimensional array used for numerical computations and does not contain labels.
# A Pandas Series is a one-dimensional labeled array that stores data along with index labels and is mainly used for data analysis.

arr = np.array([2,3,5,6,8,6])
series3 = pd.Series([2,3,4,6,8,7,6])

print("Numpy array :",arr)
print("Pandas Series :\n",series3 )
```

```python
# d) Define Series objects with individual indices
print("creating series with individual indices :\n")
pd.Series({"p":200, "q":600, "r":680, "s":500})
```

```python
# e) Access single value of a series

dic = {"a":1, "b":2, "c":3, "d":4, "e":5}
series2 = pd.Series(dic)
print("Value at index - c in 'series2' is :",series2["c"])
```

```python
# g) Usage of different methods in Matplotlib.
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,25,30]

plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()

plt.bar(x,y)
plt.title("Bar Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()

plt.scatter(x,y)
plt.title("Scatter Plot")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()


data = [2,3,3,5,6,7,8,8,9]
plt.hist(data)
plt.title("Histogram")
plt.show()


sizes = [40,30,20,10]
labels = ["A","B","C","D"]
plt.pie(sizes , labels = labels)
plt.title("Pie Chart")
plt.show()
```
