```python
# 2 (ii) a) i. Creation of Dataset using Pandas

import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name":       ["vikash", "shubham", "amit", "roshan", "aayush"],
    "Math":       [85, 90, 78, 92, 88],
    "Science":    [80, 75, 95, 85, 70],
    "English":    [88, 82, 79, 91, 84],
}

df = pd.DataFrame(data)

print("Created DataFrame using Pandas:\n",df)
print(f"\nShape   : {df.shape}")
print(f"Columns : {list(df.columns)}")  
```

```python
# 2 (ii) a) iii. Loading Datasets Using sklearn

import pandas as pd
from sklearn import datasets

iris    = datasets.load_iris()   
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_df["Species"] = [iris.target_names[t] for t in iris.target]

print("Iris Dataset")
print(f"Shape   : {iris_df.shape}")
print(f"Classes : {list(iris.target_names)}")
print("First five rows are:\n",iris_df.head())   
```

```python
# b) Mean, Median, Mode, Variance, Standard Deviation

import pandas as pd

df = pd.read_csv('airline_ticket_prices_dataset.csv')

print("Dataset:")
print(df.head())

print("\n--- Price_USD ---")
print("Mean     :", df['Price_USD'].mean())
print("Median   :", df['Price_USD'].median())
print("Mode     :", df['Price_USD'].mode()[0])
print("Variance :", df['Price_USD'].var())
print("Std Dev  :", df['Price_USD'].std())   
```

```python
# c-i) Reshaping the Data
 
import pandas as pd
import numpy as np
 
df = pd.read_csv('airline_ticket_prices_dataset.csv')

prices = df['Price_USD'].values[:12]
print('Original 1-D Price array (first 12):') 
print(prices) 
print(f'Shape: {prices.shape}')
 
reshaped = prices.reshape(4, 3)
print(f'\nReshaped to (4x3):\n{reshaped}')
```

```python
# c-ii) Filtering the Data
 
import pandas as pd
 
df = pd.read_csv('airline_ticket_prices_dataset.csv')
 
economy = df[df['Class'] == 'Economy']
print(f'Economy class tickets : {len(economy)} rows')
print(economy[['Airline', 'Origin', 'Destination', 'Price_USD']].head(3))
```

```python
# c-iii) Merging the Data
 
import pandas as pd
 
df = pd.read_csv('airline_ticket_prices_dataset.csv')
 
df2 = pd.DataFrame({
    'Airline': ['Emirates', 'Etihad', 'Saudia', 'British Airways'],
    'Country': ['UAE', 'UAE', 'Saudi Arabia', 'UK']
})
 
inner = pd.merge(df, df2, on='Airline', how='inner')
print("Inner Join:", inner.shape)
print(inner[['Airline', 'Country', 'Price_USD']].head(3)) 
```

```python
# c-iv) Handling Missing Values
 
import pandas as pd
import numpy as np
 
df = pd.read_csv('airline_ticket_prices_dataset.csv')
 
# Addding missing values
df.loc[0:5, 'Price_USD'] = np.nan
df.loc[6:8, 'Airline']   = np.nan
 
print("Missing Values:")
print(df.isnull().sum())

# Filling with mean
df['Price_USD'] = df['Price_USD'].fillna(df['Price_USD'].mean())
print("Filled with mean - Price_USD NaN left:", df['Price_USD'].isnull().sum())   
```

```python
# c-v) Min-Max Normalization
 
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
 
df = pd.read_csv('airline_ticket_prices_dataset.csv')
 
features = df[['Distance_km', 'Days_Before_Departure', 'Price_USD']]
 
print("Before Normalization:")
print(features.head(3))
 
scaler = MinMaxScaler()
scaled = scaler.fit_transform(features)
 
df_scaled = pd.DataFrame(scaled, columns=features.columns)
 
print("\nAfter Min-Max Normalization:")
print(df_scaled.head(3))   
```
